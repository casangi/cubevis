# Grid view demo (G1b)

A `GridApp` of a synthetic source: 45 baselines, a 4 x 4 grid, one toolbar for the whole grid.

Try: page with ◀ ▶, change Rows/Cols, hover a cell (probe line under the grid), box-zoom with
**Scale: common** (every cell follows) and **per cell** (only that cell), tap a cell (Python prints
the drill), **Play**, and **Save PNG** (the axes follow the scale: outer axes when common).

Every 7th cell (`ea01-ea05`, `ea02-ea04`, ...) is the odd one out.

```python
from cubevis.toolbox.gridplot.grid_app import GridApp
from cubevis.toolbox.gridplot.grid_source import SyntheticGridSource

def drill(index, label):
    print("drill", index, label)

GridApp(SyntheticGridSource(45), rows=4, cols=4, on_drill=drill).show()
```

Dark theme, scatter cells, with a simulated 50 ms render per cell (watch the status line and playback prefetch):

```python
GridApp(SyntheticGridSource(45, kind="scatter", cost=0.05), rows=3, cols=3, theme="dark").show()
```
